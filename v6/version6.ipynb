```python
import os
import json
import glob
import h5py
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from tqdm import tqdm


# ============================================================
# Config
# ============================================================

DEMO_DIR = "/content/ManiSkill/demos/PickCube-v1/motionplanning"

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

ACTION_DIM = 8
ACTION_EMBED_DIM = 128
HIDDEN_DIM = 256
WORD_DIM = 128

BATCH_SIZE = 4
EPOCHS = 100
LR = 1e-3

PAD = "<PAD>"
UNK = "<UNK>"
BOS = "<BOS>"
EOS = "<EOS>"


# ============================================================
# Instruction
# ============================================================

# 最初は1 taskだけ。
# 後でtaskごとに増やす。
INSTRUCTION = "pick up the cube"


# ============================================================
# Load trajectories
# ============================================================

def load_trajectories():

    files = sorted(
        glob.glob(
            os.path.join(
                DEMO_DIR,
                "*.h5",
            )
        )
    )

    if not files:
        raise FileNotFoundError(
            f"No h5 files found: {DEMO_DIR}"
        )

    trajectories = []

    for path in files:

        print("Loading:", path)

        with h5py.File(path, "r") as f:

            for key in sorted(f.keys()):

                if not key.startswith("traj_"):
                    continue

                actions = f[
                    f"{key}/actions"
                ][:]

                actions = torch.tensor(
                    actions,
                    dtype=torch.float32,
                )

                assert actions.ndim == 2
                assert actions.shape[1] == ACTION_DIM

                trajectories.append(actions)

    return trajectories


# ============================================================
# Vocabulary
# ============================================================

def build_vocab():

    word_to_id = {
        PAD: 0,
        UNK: 1,
        BOS: 2,
        EOS: 3,
    }

    words = INSTRUCTION.lower().split()

    for word in words:

        if word not in word_to_id:
            word_to_id[word] = len(word_to_id)

    id_to_word = {
        v: k
        for k, v in word_to_id.items()
    }

    return word_to_id, id_to_word


word_to_id, id_to_word = build_vocab()

VOCAB_SIZE = len(word_to_id)

print("Vocabulary:")
for i in range(VOCAB_SIZE):
    print(i, id_to_word[i])


# ============================================================
# Tokenize
# ============================================================

def tokenize(text):

    ids = [word_to_id[BOS]]

    for word in text.lower().split():

        ids.append(
            word_to_id.get(
                word,
                word_to_id[UNK],
            )
        )

    ids.append(word_to_id[EOS])

    return torch.tensor(
        ids,
        dtype=torch.long,
    )


# ============================================================
# Dataset
# ============================================================

class ActionLanguageDataset(Dataset):

    def __init__(self, trajectories):

        self.trajectories = trajectories

        self.tokens = tokenize(
            INSTRUCTION
        )

    def __len__(self):
        return len(self.trajectories)

    def __getitem__(self, idx):

        return {
            "actions": self.trajectories[idx],
            "tokens": self.tokens,
        }


def collate_fn(batch):

    max_len = max(
        x["actions"].shape[0]
        for x in batch
    )

    batch_size = len(batch)

    actions = torch.zeros(
        batch_size,
        max_len,
        ACTION_DIM,
        dtype=torch.float32,
    )

    lengths = torch.zeros(
        batch_size,
        dtype=torch.long,
    )

    tokens = torch.stack(
        [
            x["tokens"]
            for x in batch
        ]
    )

    for i, x in enumerate(batch):

        length = x["actions"].shape[0]

        actions[
            i,
            :length
        ] = x["actions"]

        lengths[i] = length

    return {
        "actions": actions,
        "lengths": lengths,
        "tokens": tokens,
    }


# ============================================================
# Model
# ============================================================

class ActionToLanguage(nn.Module):

    def __init__(
        self,
        action_dim,
        action_embed_dim,
        hidden_dim,
        word_dim,
        vocab_size,
    ):

        super().__init__()

        # ----------------------------------------------------
        # Action encoder
        # ----------------------------------------------------

        self.action_encoder = nn.Sequential(

            nn.Linear(
                action_dim,
                action_embed_dim,
            ),

            nn.LayerNorm(
                action_embed_dim
            ),

            nn.GELU(),
        )

        # ----------------------------------------------------
        # Action GRU
        # ----------------------------------------------------

        self.action_gru = nn.GRU(
            input_size=action_embed_dim,
            hidden_size=hidden_dim,
            batch_first=True,
        )

        # ----------------------------------------------------
        # Language embedding
        # ----------------------------------------------------

        self.word_embedding = nn.Embedding(
            vocab_size,
            word_dim,
            padding_idx=word_to_id[PAD],
        )

        # ----------------------------------------------------
        # Language decoder
        # ----------------------------------------------------

        self.language_gru = nn.GRU(
            input_size=word_dim,
            hidden_size=hidden_dim,
            batch_first=True,
        )

        self.decoder = nn.Linear(
            hidden_dim,
            vocab_size,
        )

    # ========================================================
    # Action -> latent
    # ========================================================

    def encode_action(
        self,
        actions,
        lengths,
    ):

        x = self.action_encoder(
            actions
        )

        output, hidden = self.action_gru(
            x
        )

        # ----------------------------------------------------
        # Take hidden corresponding to
        # the final valid timestep.
        # ----------------------------------------------------

        batch_size = actions.shape[0]

        last_hidden = []

        for i in range(batch_size):

            last_index = (
                lengths[i] - 1
            )

            last_hidden.append(
                output[
                    i,
                    last_index
                ]
            )

        latent = torch.stack(
            last_hidden,
            dim=0,
        )

        return latent

    # ========================================================
    # Forward
    # ========================================================

    def forward(
        self,
        actions,
        lengths,
        input_tokens,
    ):

        latent = self.encode_action(
            actions,
            lengths,
        )

        word_emb = self.word_embedding(
            input_tokens
        )

        hidden = latent.unsqueeze(0)

        output, _ = self.language_gru(
            word_emb,
            hidden,
        )

        logits = self.decoder(
            output
        )

        return latent, logits


# ============================================================
# Load
# ============================================================

print("=" * 60)
print(" AntVLA v6-alpha")
print(" Action -> Language")
print("=" * 60)

print("Device:", DEVICE)

trajectories = load_trajectories()

print(
    "Trajectories:",
    len(trajectories)
)

for i, x in enumerate(
    trajectories
):

    print(
        f"  traj_{i}:",
        tuple(x.shape)
    )


# ============================================================
# Dataset
# ============================================================

dataset = ActionLanguageDataset(
    trajectories
)

loader = DataLoader(
    dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    collate_fn=collate_fn,
)


# ============================================================
# Model
# ============================================================

model = ActionToLanguage(
    action_dim=ACTION_DIM,
    action_embed_dim=ACTION_EMBED_DIM,
    hidden_dim=HIDDEN_DIM,
    word_dim=WORD_DIM,
    vocab_size=VOCAB_SIZE,
).to(DEVICE)


parameter_count = sum(
    p.numel()
    for p in model.parameters()
)

print(
    "Parameters:",
    f"{parameter_count:,}"
)


# ============================================================
# Optimizer
# ============================================================

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=LR,
)

criterion = nn.CrossEntropyLoss(
    ignore_index=word_to_id[PAD]
)


# ============================================================
# Training
# ============================================================

print("\nTraining...")

for epoch in range(EPOCHS):

    model.train()

    total_loss = 0.0

    pbar = tqdm(
        loader,
        desc=f"Epoch {epoch + 1}/{EPOCHS}",
    )

    for batch in pbar:

        actions = batch[
            "actions"
        ].to(DEVICE)

        lengths = batch[
            "lengths"
        ]

        target = batch[
            "tokens"
        ].to(DEVICE)

        # ----------------------------------------------------
        # Teacher forcing
        # ----------------------------------------------------

        input_tokens = target[:, :-1]

        target_tokens = target[:, 1:]

        # ----------------------------------------------------
        # Forward
        # ----------------------------------------------------

        latent, logits = model(
            actions,
            lengths,
            input_tokens,
        )

        loss = criterion(
            logits.reshape(
                -1,
                VOCAB_SIZE,
            ),
            target_tokens.reshape(
                -1
            ),
        )

        # ----------------------------------------------------
        # Backprop
        # ----------------------------------------------------

        optimizer.zero_grad(
            set_to_none=True
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            1.0,
        )

        optimizer.step()

        total_loss += loss.item()

        pbar.set_postfix(
            loss=f"{loss.item():.5f}"
        )

    avg_loss = (
        total_loss
        / len(loader)
    )

    print(
        f"Epoch {epoch + 1}: "
        f"loss={avg_loss:.6f}"
    )


# ============================================================
# Greedy generation
# ============================================================

def generate(
    model,
    actions,
    max_len=20,
):

    model.eval()

    with torch.no_grad():

        actions = actions.unsqueeze(0).to(
            DEVICE
        )

        lengths = torch.tensor(
            [actions.shape[1]],
            dtype=torch.long,
        )

        latent = model.encode_action(
            actions,
            lengths,
        )

        hidden = latent.unsqueeze(0)

        current = torch.tensor(
            [[word_to_id[BOS]]],
            dtype=torch.long,
            device=DEVICE,
        )

        result = []

        for _ in range(max_len):

            emb = model.word_embedding(
                current
            )

            output, hidden = (
                model.language_gru(
                    emb,
                    hidden,
                )
            )

            logits = model.decoder(
                output[:, -1]
            )

            next_id = logits.argmax(
                dim=-1
            ).item()

            if next_id == word_to_id[EOS]:
                break

            if next_id != word_to_id[PAD]:
                result.append(
                    id_to_word[next_id]
                )

            current = torch.tensor(
                [[next_id]],
                dtype=torch.long,
                device=DEVICE,
            )

        return " ".join(result)


# ============================================================
# Test
# ============================================================

print("\nGeneration test:")

for i in range(
    min(10, len(trajectories))
):

    text = generate(
        model,
        trajectories[i],
    )

    print(
        f"traj_{i}: {text}"
    )


# ============================================================
# Save
# ============================================================

os.makedirs(
    "/content/antvla-v6/checkpoints",
    exist_ok=True,
)

save_path = (
    "/content/antvla-v6/checkpoints/"
    "action_to_language_gru.pt"
)

torch.save(
    {
        "model": model.state_dict(),
        "action_dim": ACTION_DIM,
        "action_embed_dim": ACTION_EMBED_DIM,
        "hidden_dim": HIDDEN_DIM,
        "word_dim": WORD_DIM,
        "vocab_size": VOCAB_SIZE,
        "word_to_id": word_to_id,
        "id_to_word": id_to_word,
        "instruction": INSTRUCTION,
    },
    save_path,
)

print("\nSaved:", save_path)
print("=" * 60)
print("ACTION -> LANGUAGE: DONE")
print("=" * 60)
```

```python
# Action -> Language GRU : 100 trajectories
# /content/antvla-v6

import os
import h5py
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

H5_PATH = "/content/ManiSkill/demos/PickCube-v1/motionplanning/pickcube_100.h5"
SAVE_PATH = "/content/antvla-v6/checkpoints/action_to_language_gru_100.pt"

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

VOCAB = {
    "<PAD>": 0,
    "<UNK>": 1,
    "<BOS>": 2,
    "<EOS>": 3,
    "pick": 4,
    "up": 5,
    "the": 6,
    "cube": 7,
}

ID2WORD = {v: k for k, v in VOCAB.items()}


class ActionDataset(Dataset):
    def __init__(self, path):
        self.data = []

        with h5py.File(path, "r") as f:
            trajs = sorted(
                [k for k in f.keys() if k.startswith("traj_")],
                key=lambda x: int(x.split("_")[1])
            )

            for name in trajs:
                actions = f[name]["actions"][:]
                self.data.append(torch.tensor(actions, dtype=torch.float32))

        print("Trajectories:", len(self.data))

    def __len__(self):
        return len(self.data)

    def __getitem__(self, i):
        return self.data[i]


class ActionToLanguageGRU(nn.Module):
    def __init__(
        self,
        action_dim=8,
        hidden_dim=256,
        vocab_size=len(VOCAB),
    ):
        super().__init__()

        self.encoder = nn.Sequential(
            nn.Linear(action_dim, 128),
            nn.ReLU(),
            nn.Linear(128, hidden_dim),
        )

        self.gru = nn.GRU(
            hidden_dim,
            hidden_dim,
            batch_first=True,
        )

        self.embedding = nn.Embedding(vocab_size, hidden_dim)

        self.decoder = nn.Linear(hidden_dim, vocab_size)

    def forward(self, actions, tokens):
        x = self.encoder(actions)

        _, h = self.gru(x)

        # BOS -> words -> EOS
        emb = self.embedding(tokens)

        h_seq, _ = self.gru(
            emb,
            h
        )

        logits = self.decoder(h_seq)

        return logits


dataset = ActionDataset(H5_PATH)

# 全trajectoryをpaddingせず1本ずつ処理
loader = DataLoader(
    dataset,
    batch_size=1,
    shuffle=True,
)

model = ActionToLanguageGRU().to(device)

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=1e-3,
)

criterion = nn.CrossEntropyLoss()

# "pick up the cube"
target = torch.tensor(
    [[
        VOCAB["<BOS>"],
        VOCAB["pick"],
        VOCAB["up"],
        VOCAB["the"],
        VOCAB["cube"],
        VOCAB["<EOS>"],
    ]],
    device=device,
)

print("=" * 60)
print("AntVLA v6-alpha")
print("Action -> Language / GRU")
print("=" * 60)
print("Device:", device)
print("Parameters:", sum(p.numel() for p in model.parameters()))

for epoch in range(1, 101):

    model.train()
    total_loss = 0.0

    for actions in loader:

        actions = actions.to(device)

        # token input:
        # BOS pick up the cube
        inp = target[:, :-1]

        # prediction:
        # pick up the cube EOS
        labels = target[:, 1:]

        logits = model(actions, inp)

        loss = criterion(
            logits.reshape(-1, len(VOCAB)),
            labels.reshape(-1),
        )

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    avg = total_loss / len(loader)

    if epoch == 1 or epoch % 10 == 0:
        print(f"Epoch {epoch:3d}: loss={avg:.6f}")


os.makedirs(os.path.dirname(SAVE_PATH), exist_ok=True)

torch.save(
    {
        "model": model.state_dict(),
        "vocab": VOCAB,
    },
    SAVE_PATH,
)

print()
print("Saved:", SAVE_PATH)
```

```python
import os
import h5py
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

PICK = "/content/ManiSkill/demos/PickCube-v1/motionplanning/pickcube_100.h5"
STACK = "/content/ManiSkill/demos/StackCube-v1/motionplanning/stackcube_100.h5"

VOCAB = {
    "<PAD>": 0,
    "<UNK>": 1,
    "<BOS>": 2,
    "<EOS>": 3,
    "pick": 4,
    "up": 5,
    "the": 6,
    "cube": 7,
    "stack": 8,
    "cubes": 9,
}

ID2WORD = {v: k for k, v in VOCAB.items()}


class ActionLanguageDataset(Dataset):
    def __init__(self):
        self.data = []

        for path, sentence in [
            (PICK, ["pick", "up", "the", "cube"]),
            (STACK, ["stack", "the", "cubes"]),
        ]:
            with h5py.File(path, "r") as f:
                trajs = sorted(
                    [k for k in f.keys() if k.startswith("traj_")],
                    key=lambda x: int(x.split("_")[1])
                )

                for name in trajs:
                    actions = torch.tensor(
                        f[name]["actions"][:],
                        dtype=torch.float32
                    )

                    tokens = torch.tensor(
                        [VOCAB["<BOS>"]] +
                        [VOCAB[w] for w in sentence] +
                        [VOCAB["<EOS>"]],
                        dtype=torch.long
                    )

                    self.data.append((actions, tokens))

        print("Total trajectories:", len(self.data))

    def __len__(self):
        return len(self.data)

    def __getitem__(self, i):
        return self.data[i]


class ActionToLanguageGRU(nn.Module):
    def __init__(self, action_dim=8, hidden_dim=256):
        super().__init__()

        self.encoder = nn.Sequential(
            nn.Linear(action_dim, 128),
            nn.ReLU(),
            nn.Linear(128, hidden_dim),
        )

        self.action_gru = nn.GRU(
            hidden_dim,
            hidden_dim,
            batch_first=True
        )

        self.embedding = nn.Embedding(
            len(VOCAB),
            hidden_dim
        )

        self.decoder_gru = nn.GRU(
            hidden_dim,
            hidden_dim,
            batch_first=True
        )

        self.decoder = nn.Linear(
            hidden_dim,
            len(VOCAB)
        )

    def encode_actions(self, actions):
        x = self.encoder(actions)
        _, h = self.action_gru(x)
        return h

    def forward(self, actions, tokens):
        h = self.encode_actions(actions)

        x = self.embedding(tokens)

        y, _ = self.decoder_gru(x, h)

        return self.decoder(y)


dataset = ActionLanguageDataset()

loader = DataLoader(
    dataset,
    batch_size=1,
    shuffle=True
)

model = ActionToLanguageGRU().to(device)

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=1e-3
)

criterion = nn.CrossEntropyLoss()

print("=" * 60)
print("AntVLA v6-alpha")
print("200 trajectories")
print("PickCube + StackCube")
print("=" * 60)

print("Device:", device)
print(
    "Parameters:",
    sum(p.numel() for p in model.parameters())
)


for epoch in range(1, 101):

    model.train()
    total_loss = 0.0

    for actions, tokens in loader:

        actions = actions.to(device)
        tokens = tokens.to(device)

        inp = tokens[:, :-1]
        labels = tokens[:, 1:]

        logits = model(actions, inp)

        loss = criterion(
            logits.reshape(-1, len(VOCAB)),
            labels.reshape(-1)
        )

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    avg = total_loss / len(loader)

    if epoch == 1 or epoch % 10 == 0:
        print(
            f"Epoch {epoch:3d}: "
            f"loss={avg:.6f}"
        )


SAVE = "/content/antvla-v6/checkpoints/action_to_language_gru_200.pt"

os.makedirs(os.path.dirname(SAVE), exist_ok=True)

torch.save(
    {
        "model": model.state_dict(),
        "vocab": VOCAB,
    },
    SAVE
)

print()
print("Saved:", SAVE)
```

```python
import h5py
import torch
import torch.nn as nn

# ============================================================
# AntVLA v6-alpha
# Action -> Language
# Evaluation: PickCube + StackCube
# ============================================================

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

CKPT = "/content/antvla-v6/checkpoints/action_to_language_gru_200.pt"

PICK = "/content/ManiSkill/demos/PickCube-v1/motionplanning/pickcube_100.h5"
STACK = "/content/ManiSkill/demos/StackCube-v1/motionplanning/stackcube_100.h5"


# ------------------------------------------------------------
# Vocabulary
# ------------------------------------------------------------

VOCAB = {
    "<PAD>": 0,
    "<UNK>": 1,
    "<BOS>": 2,
    "<EOS>": 3,
    "pick": 4,
    "up": 5,
    "the": 6,
    "cube": 7,
    "stack": 8,
    "cubes": 9,
}

ID2WORD = {v: k for k, v in VOCAB.items()}


# ------------------------------------------------------------
# Model
# ------------------------------------------------------------

class ActionToLanguageGRU(nn.Module):

    def __init__(self, action_dim=8, hidden_dim=256):
        super().__init__()

        self.encoder = nn.Sequential(
            nn.Linear(action_dim, 128),
            nn.ReLU(),
            nn.Linear(128, hidden_dim),
        )

        self.action_gru = nn.GRU(
            hidden_dim,
            hidden_dim,
            batch_first=True
        )

        self.embedding = nn.Embedding(
            len(VOCAB),
            hidden_dim
        )

        self.decoder_gru = nn.GRU(
            hidden_dim,
            hidden_dim,
            batch_first=True
        )

        self.decoder = nn.Linear(
            hidden_dim,
            len(VOCAB)
        )

    def encode_actions(self, actions):

        x = self.encoder(actions)

        _, h = self.action_gru(x)

        return h

    def decode(self, h, max_len=10):

        token = torch.tensor(
            [[VOCAB["<BOS>"]]],
            device=DEVICE
        )

        words = []

        for _ in range(max_len):

            x = self.embedding(token)

            y, h = self.decoder_gru(x, h)

            logits = self.decoder(y[:, -1])

            next_token = logits.argmax(dim=-1)

            idx = next_token.item()

            if idx == VOCAB["<EOS>"]:
                break

            if idx not in (
                VOCAB["<PAD>"],
                VOCAB["<BOS>"]
            ):
                words.append(ID2WORD.get(idx, "<UNK>"))

            token = next_token.unsqueeze(0)

        return " ".join(words)


# ------------------------------------------------------------
# Load
# ------------------------------------------------------------

model = ActionToLanguageGRU().to(DEVICE)

checkpoint = torch.load(
    CKPT,
    map_location=DEVICE
)

model.load_state_dict(checkpoint["model"])
model.eval()

print("=" * 60)
print("AntVLA v6-alpha")
print("Action -> Language Evaluation")
print("=" * 60)
print("Device:", DEVICE)


# ------------------------------------------------------------
# Evaluation
# ------------------------------------------------------------

def evaluate(path, expected):

    results = []

    with h5py.File(path, "r") as f:

        trajs = sorted(
            [
                k for k in f.keys()
                if k.startswith("traj_")
            ],
            key=lambda x: int(x.split("_")[1])
        )

        for name in trajs:

            actions = torch.tensor(
                f[name]["actions"][:],
                dtype=torch.float32,
                device=DEVICE
            ).unsqueeze(0)

            with torch.no_grad():

                h = model.encode_actions(actions)

                prediction = model.decode(h)

            correct = prediction == expected

            results.append(correct)

            print(
                f"{name:8s} "
                f"→ {prediction}"
                + (" ✓" if correct else " ✗")
            )

    accuracy = sum(results) / len(results)

    print()
    print(
        f"Accuracy: "
        f"{sum(results)}/{len(results)} "
        f"({accuracy * 100:.1f}%)"
    )

    return accuracy


print()
print("=" * 60)
print("PickCube")
print("=" * 60)

pick_acc = evaluate(
    PICK,
    "pick up the cube"
)


print()
print("=" * 60)
print("StackCube")
print("=" * 60)

stack_acc = evaluate(
    STACK,
    "stack the cubes"
)


print()
print("=" * 60)
print("FINAL")
print("=" * 60)

print(
    f"PickCube : {pick_acc * 100:.1f}%"
)

print(
    f"StackCube: {stack_acc * 100:.1f}%"
)

print(
    f"Overall  : "
    f"{(pick_acc + stack_acc) / 2 * 100:.1f}%"
)
```

```python
import h5py
import torch
import torch.nn as nn

# ============================================================
# AntVLA v6-alpha
# UNSEEN TRAJECTORY TEST
# ============================================================

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

CKPT = "/content/antvla-v6/checkpoints/action_to_language_gru_200.pt"

# 追加生成した未知 trajectory
PICK_TEST = "/content/ManiSkill/demos/PickCube-v1/motionplanning/pickcube_test_100.h5"
STACK_TEST = "/content/ManiSkill/demos/StackCube-v1/motionplanning/stackcube_test_100.h5"


# ------------------------------------------------------------
# Vocabulary
# ------------------------------------------------------------

VOCAB = {
    "<PAD>": 0,
    "<UNK>": 1,
    "<BOS>": 2,
    "<EOS>": 3,
    "pick": 4,
    "up": 5,
    "the": 6,
    "cube": 7,
    "stack": 8,
    "cubes": 9,
}

ID2WORD = {v: k for k, v in VOCAB.items()}


# ------------------------------------------------------------
# Model
# ------------------------------------------------------------

class ActionToLanguageGRU(nn.Module):

    def __init__(self, action_dim=8, hidden_dim=256):
        super().__init__()

        self.encoder = nn.Sequential(
            nn.Linear(action_dim, 128),
            nn.ReLU(),
            nn.Linear(128, hidden_dim),
        )

        self.action_gru = nn.GRU(
            hidden_dim,
            hidden_dim,
            batch_first=True
        )

        self.embedding = nn.Embedding(
            len(VOCAB),
            hidden_dim
        )

        self.decoder_gru = nn.GRU(
            hidden_dim,
            hidden_dim,
            batch_first=True
        )

        self.decoder = nn.Linear(
            hidden_dim,
            len(VOCAB)
        )

    def encode_actions(self, actions):

        x = self.encoder(actions)

        _, h = self.action_gru(x)

        return h

    def decode(self, h, max_len=10):

        token = torch.tensor(
            [[VOCAB["<BOS>"]]],
            device=DEVICE
        )

        words = []

        for _ in range(max_len):

            x = self.embedding(token)

            y, h = self.decoder_gru(x, h)

            logits = self.decoder(y[:, -1])

            next_token = logits.argmax(dim=-1)

            idx = next_token.item()

            if idx == VOCAB["<EOS>"]:
                break

            if idx not in (
                VOCAB["<PAD>"],
                VOCAB["<BOS>"]
            ):
                words.append(
                    ID2WORD.get(idx, "<UNK>")
                )

            token = next_token.unsqueeze(0)

        return " ".join(words)


# ------------------------------------------------------------
# Load trained model
# ------------------------------------------------------------

model = ActionToLanguageGRU().to(DEVICE)

checkpoint = torch.load(
    CKPT,
    map_location=DEVICE
)

model.load_state_dict(checkpoint["model"])

model.eval()

print("=" * 60)
print("AntVLA v6-alpha")
print("UNSEEN TRAJECTORY TEST")
print("=" * 60)
print("Device:", DEVICE)
print("Checkpoint:", CKPT)


# ------------------------------------------------------------
# Evaluation
# ------------------------------------------------------------

def evaluate(path, expected):

    results = []

    with h5py.File(path, "r") as f:

        trajs = sorted(
            [
                k for k in f.keys()
                if k.startswith("traj_")
            ],
            key=lambda x: int(x.split("_")[1])
        )

        print()
        print("File:", path)
        print("Trajectories:", len(trajs))
        print("-" * 60)

        for name in trajs:

            actions = torch.tensor(
                f[name]["actions"][:],
                dtype=torch.float32,
                device=DEVICE
            ).unsqueeze(0)

            with torch.no_grad():

                h = model.encode_actions(actions)

                prediction = model.decode(h)

            correct = prediction == expected

            results.append(correct)

            print(
                f"{name:12s} → "
                f"{prediction:25s}"
                + (" ✓" if correct else " ✗")
            )

    correct_count = sum(results)
    total_count = len(results)

    accuracy = (
        correct_count / total_count
        if total_count > 0
        else 0.0
    )

    print()
    print(
        f"Accuracy: "
        f"{correct_count}/{total_count} "
        f"({accuracy * 100:.1f}%)"
    )

    return correct_count, total_count


# ------------------------------------------------------------
# PickCube TEST
# ------------------------------------------------------------

print()
print("=" * 60)
print("PickCube — UNSEEN TEST")
print("=" * 60)

pick_correct, pick_total = evaluate(
    PICK_TEST,
    "pick up the cube"
)


# ------------------------------------------------------------
# StackCube TEST
# ------------------------------------------------------------

print()
print("=" * 60)
print("StackCube — UNSEEN TEST")
print("=" * 60)

stack_correct, stack_total = evaluate(
    STACK_TEST,
    "stack the cubes"
)


# ------------------------------------------------------------
# FINAL
# ------------------------------------------------------------

total_correct = pick_correct + stack_correct
total_total = pick_total + stack_total

overall = (
    total_correct / total_total
    if total_total > 0
    else 0.0
)

print()
print("=" * 60)
print("FINAL — UNSEEN TRAJECTORY TEST")
print("=" * 60)

print(
    f"PickCube : "
    f"{pick_correct}/{pick_total} "
    f"({pick_correct / pick_total * 100:.1f}%)"
)

print(
    f"StackCube: "
    f"{stack_correct}/{stack_total} "
    f"({stack_correct / stack_total * 100:.1f}%)"
)

print(
    f"Overall  : "
    f"{total_correct}/{total_total} "
    f"({overall * 100:.1f}%)"
)
```
